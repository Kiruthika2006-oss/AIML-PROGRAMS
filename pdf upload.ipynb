```
# Import the AutoTokenizer from Hugging Face
from transformers import AutoTokenizer

# 1. Define a representative sentence for the RAG (Retrieval-Augmented Generation) concept
rag_sentence = "Retrieval-Augmented Generation (RAG) improves LLM responses by retrieving relevant facts from an external knowledge base before generating a final answer."
print(f"Original Sentence:\n{rag_sentence}\n")

# 2. Load a standard pre-trained tokenizer (using BERT-base-uncased as an example)
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# 3. Tokenize the sentence
tokens = tokenizer.tokenize(rag_sentence)
print(f"Tokens:\n{tokens}\n")

# 4. Map tokens to their corresponding input IDs (token IDs)
input_ids = tokenizer.convert_tokens_to_ids(tokens)
print(f"Token IDs:\n{input_ids}")
```

```
# Define the new target text
target_text = "## Resources"

# Tokenize the target text
target_tokens = tokenizer.tokenize(target_text)
print(f"Tokens for '{target_text}':\n{target_tokens}\n")

# Map to token IDs
target_token_ids = tokenizer.convert_tokens_to_ids(target_tokens)
print(f"Token IDs:\n{target_token_ids}")
```

```
import torch
from transformers import AutoModel

# 1. Load the pre-trained BERT model
model = AutoModel.from_pretrained("bert-base-uncased")

# 2. Convert target_token_ids to a PyTorch tensor and add a batch dimension
target_ids_tensor = torch.tensor([target_token_ids])

# 3. Generate embeddings without calculating gradients
with torch.no_grad():
    outputs = model(target_ids_tensor)

# 4. Extract the last hidden states
target_embeddings = outputs.last_hidden_state[0]

# 5. Convert to fully-expanded numerical list and display
target_embeddings_list = target_embeddings.tolist()
for token_name, embedding_vector in zip(target_tokens, target_embeddings_list):
    print(f"Token: {token_name:<12} | Dimensions: {len(embedding_vector)}")
    print(f"Embedding: {embedding_vector}\n")
```
